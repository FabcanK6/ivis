Runtime → Change runtime type → **T4 GPU**. Training `bert-base-uncased` on 20k examples for 4 epochs takes roughly 10–15 minutes on a T4.

```python
!python -m ivis.data.generate --out data --n-train 20000 --n-val 2000 --n-test 2000
```

```python
!python -m ivis.train --model bert-base-uncased --out models/ivis-bert --epochs 4 --batch-size 32 --fp16
```

```python
!python -m ivis.evaluate --data data/test.jsonl --model models/ivis-bert --report reports/bert_test.json
!python -m ivis.evaluate --data data/test.jsonl --backend rules --report reports/rules_test.json --errors 0
```

```python
from ivis.predict import load_parser
import json
parser = load_parser("models/ivis-bert")
spec = parser.parse("Can we get a chart of the top 10 sites with the most open queries in Germany over the last 30 days?")
print(json.dumps({k: spec[k] for k in ["title", "powerbi_visual", "field_wells", "filters", "time_filter", "top_n", "confidence"]}, indent=2))
```
